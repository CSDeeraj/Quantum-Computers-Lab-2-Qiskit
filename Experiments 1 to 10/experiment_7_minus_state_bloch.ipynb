### CODE

The following cell implements the experiment using Qiskit.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.z(0)
state = Statevector.from_instruction(qc)

print("Circuit:")
display(qc.draw("mpl"))
print("Statevector:", state)
print("Expected state: |-> = (|0> - |1>)/sqrt(2)")

plot_bloch_multivector(state)
plt.show()
```

### OUTPUT / OBSERVATION
Run the code cell above to display the quantum circuit, statevector, and Bloch Sphere.

### RESULT
H prepares |+⟩ and Z changes it to |−⟩ = (|0⟩ − |1⟩)/√2. The Bloch vector points along −x on the equator.
