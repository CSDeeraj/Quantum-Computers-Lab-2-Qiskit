### CODE

The following cell implements the experiment using Qiskit.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
qc = QuantumCircuit(1)
qc.h(0)
state = Statevector.from_instruction(qc)

print("Statevector:", state)
print("Expected state: |+> = (|0> + |1>)/sqrt(2)")

plot_bloch_multivector(state)
plt.show()
```

### OUTPUT / OBSERVATION
Run the code cell above to display the quantum circuit, statevector, and Bloch Sphere.

### RESULT
The Hadamard gate creates |+⟩ = (|0⟩ + |1⟩)/√2, represented by the +x direction on the equator.
