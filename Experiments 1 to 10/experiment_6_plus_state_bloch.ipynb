### PROCEDURE

2. Create a single-qubit quantum circuit initialized in |0⟩.
3. Apply the required quantum gate(s).
4. Generate the statevector.
5. Visualize the state using the Bloch Sphere.
6. Observe and analyze the resulting state.

### PROGRAM CODE

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
qc = QuantumCircuit(1)
qc.h(0)
state = Statevector.from_instruction(qc)
print("Statevector:", state)
print("Expected state: |+> = (|0> + |1>)/sqrt(2)")
plot_bloch_multivector(state)
plt.show()
```

### OUTPUT / OBSERVATION
The code above generates the quantum circuit, statevector, and Bloch Sphere visualization.

### RESULT
The Hadamard gate produces the |+⟩ state on the +x direction of the equator.
