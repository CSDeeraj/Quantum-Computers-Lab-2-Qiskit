### CODE

The following cell implements the experiment using Qiskit.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
qc = QuantumCircuit(1)
qc.ry(np.pi / 3, 0)
qc.rz(np.pi / 4, 0)

state = Statevector.from_instruction(qc)

print("Circuit:")
display(qc.draw("mpl"))
print("\nStatevector:")
print(state)

print("\nBloch Sphere representation:")
plot_bloch_multivector(state)
plt.show()
```

### OUTPUT / OBSERVATION
Run the code cell above to display the quantum circuit, statevector, and Bloch Sphere.
