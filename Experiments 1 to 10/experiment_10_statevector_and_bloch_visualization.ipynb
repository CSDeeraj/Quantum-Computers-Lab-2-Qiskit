### PROCEDURE

3. Apply the required quantum gate(s).
4. Generate the statevector.
5. Visualize the state using the Bloch Sphere.
6. Observe and analyze the resulting state.

### PROGRAM CODE

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
qc = QuantumCircuit(1)
qc.ry(np.pi / 3, 0)
qc.rz(np.pi / 4, 0)
state = Statevector.from_instruction(qc)

print("Circuit:")
display(qc.draw("mpl"))
print("\nStatevector:")
print(state)
print("\nBloch Sphere representation:")
plot_bloch_multivector(state)
plt.show()
```

### OUTPUT / OBSERVATION
The code above generates the quantum circuit, statevector, and Bloch Sphere visualization.

### RESULT
The generated pure single-qubit state is represented by a point on the surface of the Bloch Sphere.
