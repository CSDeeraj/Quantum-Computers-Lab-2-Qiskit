### PROCEDURE
1. Import the required Qiskit libraries and visualization modules.
2. Create a single-qubit quantum circuit initialized in |0⟩.
3. Apply the required quantum gate(s).
4. Generate the statevector.
5. Visualize the state using the Bloch Sphere.
6. Observe and analyze the resulting state.

### PROGRAM CODE

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
circuits = {
    "Pauli-X": QuantumCircuit(1),
    "Pauli-Y": QuantumCircuit(1),
    "Pauli-Z": QuantumCircuit(1)
}
circuits["Pauli-X"].x(0)
circuits["Pauli-Y"].y(0)
circuits["Pauli-Z"].z(0)

for name, circuit in circuits.items():
    state = Statevector.from_instruction(circuit)
    print(f"\n{name}")
    print("Statevector:", state)
    display(circuit.draw("mpl"))
    plot_bloch_multivector(state)
    plt.show()
```

### OUTPUT / OBSERVATION
The code above generates the quantum circuit, statevector, and Bloch Sphere visualization.

### RESULT
From |0⟩, X gives |1⟩, Y gives i|1⟩ (same physical Bloch point as |1⟩), and Z leaves |0⟩ unchanged.
