### CODE

The following cell implements the experiment using Qiskit.

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
```

```python
circuits = {
    "Pauli-X": QuantumCircuit(1),
    "Pauli-Y": QuantumCircuit(1),
    "Pauli-Z": QuantumCircuit(1),
}

circuits["Pauli-X"].x(0)
circuits["Pauli-Y"].y(0)
circuits["Pauli-Z"].z(0)

for name, circuit in circuits.items():
    state = Statevector.from_instruction(circuit)
    print(f"\n{name}")
    print("Circuit:")
    display(circuit.draw("mpl"))
    print("Statevector:", state)
    plot_bloch_multivector(state)
    plt.show()
```

### OUTPUT / OBSERVATION
Run the code cell above to display the quantum circuit, statevector, and Bloch Sphere.

### RESULT
From |0⟩: X produces |1⟩, Y produces i|1⟩ (same physical Bloch point as |1⟩), and Z leaves |0⟩ unchanged. The comparison highlights how the gates transform the state.
